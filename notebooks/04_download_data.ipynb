# Download Data from Zenodo to Google Drive

Run this notebook **once** in Google Colab to download the train/val/test parquet files
from Zenodo and save them permanently to your Google Drive.

After this notebook completes, use the printed `BASE_PATH` in all model notebooks.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os
import urllib.request

# Permanent folder on your Google Drive where the files will be saved
BASE_PATH     = '/content/drive/MyDrive/flight_data'
ZENODO_RECORD = 'https://zenodo.org/record/20469826/files'

os.makedirs(BASE_PATH, exist_ok=True)

for name in ['train.parquet', 'val.parquet', 'test.parquet']:
    out = f'{BASE_PATH}/{name}'
    if os.path.exists(out):
        size_gb = os.path.getsize(out) / (1024**3)
        print(f'{name} already exists ({size_gb:.2f} GB) — skipping')
    else:
        print(f'Downloading {name}...')
        !wget -q --show-progress "{ZENODO_RECORD}/{name}?download=1" -O "{out}"
        size_gb = os.path.getsize(out) / (1024**3)
        print(f'  Saved: {out} ({size_gb:.2f} GB)')

print()
print('All files ready.')
print(f'Set BASE_PATH = "{BASE_PATH}" in all model notebooks.')
```
