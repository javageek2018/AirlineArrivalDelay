# Data Size Check
Run this notebook in Google Colab to check the sizes of the train/val/test parquet folders on Drive.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os

BASE_PATH = '/content/drive/MyDrive/OMDS Capstone/Data/flights_all_features_encoded'

splits = ['train.parquet', 'val.parquet', 'test.parquet']

total_bytes = 0
print(f'{'File':<20} {'Size (GB)':>10} {'# Part Files':>14}')
print('-' * 48)

for name in splits:
    path = os.path.join(BASE_PATH, name)
    file_bytes = 0
    n_files = 0
    for dirpath, dirnames, filenames in os.walk(path):
        for f in filenames:
            file_bytes += os.path.getsize(os.path.join(dirpath, f))
            n_files += 1
    total_bytes += file_bytes
    print(f'{name:<20} {file_bytes / (1024**3):>10.2f} {n_files:>14}')

print('-' * 48)
print(f'{'TOTAL':<20} {total_bytes / (1024**3):>10.2f}')
print()
print(f'Zenodo limit  : 50 GB  -> {"OK" if total_bytes / (1024**3) < 50 else "EXCEEDS LIMIT"}')
print(f'Kaggle limit  : 100 GB -> {"OK" if total_bytes / (1024**3) < 100 else "EXCEEDS LIMIT"}')
```
